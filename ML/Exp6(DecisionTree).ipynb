```python
import pandas as pd
df=pd.read_csv('weather_data.csv')
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for column in df.columns:
    df[column] = le.fit_transform(df[column])
```

```python
# Step 3: Split data into features (X) and target (y)
X = df.drop('play golf', axis=1)
y = df['play golf']
```

```python

from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(criterion='entropy')
model.fit(X, y)
```

```python
from matplotlib import pyplot as plt
from sklearn import tree

# Step 5: Visualize Decision Tree
plt.figure(figsize=(12,8))
tree.plot_tree(model, feature_names=X.columns, class_names=['No','Yes'], filled=True)
plt.show()
```

```python
# Step 6: Predict new data
# Example: sunny, cool, high, true
# You must encode same way → (check mapping by printing df)

sample = [[2, 0, 0, 1]]  # Example encoded values
prediction = model.predict(sample)

print("Prediction:", "Yes" if prediction[0] == 1 else "No")
```
