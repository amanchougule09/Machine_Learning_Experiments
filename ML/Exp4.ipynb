```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head()
```

```python
class_0 = df[df['Purchased'] == 0]
class_1 = df[df['Purchased'] == 1]

plt.scatter(class_0['Age'],class_0['EstimatedSalary'], color='red')
plt.scatter(class_1['Age'],class_1['EstimatedSalary'], color='blue')
```

```python
X = df[['Age', 'EstimatedSalary']]
y = df['Purchased']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.25, random_state= 18 )
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train_scaled,y_train)

q0 = [25,40000]
model.predict(scaler.transform([q0]))
```

```python
for i in range(1,10,2):
    model = KNeighborsClassifier(n_neighbors = i)
    model.fit(X_train_scaled , y_train)
    score1 =model.score(X_train_scaled , y_train)
    print("N",i,"Score",score1)
```

```python
final_model =KNeighborsClassifier(n_neighbors=1)
X_scaled =scaler.fit_transform(X)
final_model.fit(X_scaled,y)
final_model.score(X_scaled,y)
```
