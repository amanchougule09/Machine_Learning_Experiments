```python
import numpy as np
```

```python


# AND gate inputs
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Taking weights and bias from user
w1 = float(input("Enter weight w1: "))
w2 = float(input("Enter weight w2: "))
b = float(input("Enter bias: "))

weights = np.array([w1, w2])

print("\nResults using Sigmoid Function:\n")

# Processing each input
for i in range(len(X)):
    z = np.dot(X[i], weights) + b   # Weighted sum
    output = sigmoid(z)             # Sigmoid output
    
    print(f"Input: {X[i]} -> Output: {output:.4f}")
```
